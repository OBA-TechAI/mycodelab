```python
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------------------------------
# 1️⃣ LOAD ADMISSIONS DATA
# -------------------------------------------------------
# Update path to your admissions.csv file
f_Path = "/content/admissions.csv"
f_Path1 = "/content/admissions1.csv"

# Load only required columns for efficiency for MIMIC-III
df = pd.read_csv(
    f_Path,
    usecols=["SUBJECT_ID", "HADM_ID", "ADMITTIME", "DISCHTIME"]
)

# Load only required columns for efficiency for MIMIC-IV
df1 = pd.read_csv(
    f_Path1,
    usecols=["subject_id", "hadm_id", "admittime", "dischtime"] # Corrected usecols
)

# 2️⃣ PREPARE TIME VARIABLES

# Convert to datetime format for MIMIC-III
df["ADMITTIME"] = pd.to_datetime(df["ADMITTIME"])
df["DISCHTIME"] = pd.to_datetime(df["DISCHTIME"])

# Convert to datetime format for MIMIC-IV
df1["admittime"] = pd.to_datetime(df1["admittime"]) # Corrected DataFrame reference
df1["dischtime"] = pd.to_datetime(df1["dischtime"]) # Corrected DataFrame reference

# Sort by patient and admission time
df = df.sort_values(["SUBJECT_ID", "ADMITTIME"])
df1 = df1.sort_values(["subject_id", "admittime"])

# -------------------------------------------------------
# 3️⃣ CALCULATE TOTAL ADMISSIONS PER PATIENT
# -------------------------------------------------------
admission_counts = (
    df.groupby("SUBJECT_ID")["HADM_ID"]
      .nunique()
      .reset_index(name="T_hosp_visits")
)

admission_counts1 = (
    df1.groupby("subject_id")["hadm_id"] # Corrected DataFrame reference
      .nunique()
      .reset_index(name="T_hosp1_visits")
)

# 4️⃣ CALCULATE 30-DAY READMISSION
# -------------------------------------------------------
# Get next admission per patient for MIMIC-III
df["next_admittime"] = df.groupby("SUBJECT_ID")["ADMITTIME"].shift(-1)

# Get next admission per patient for MIMIC-IV
df1["next_admittime1"] = df1.groupby("subject_id")["admittime"].shift(-1) # Corrected typo 'amittime' to 'admittime'

# Calculate difference in days between discharge and next admission for MIMIC-III
df["days_to_readmit"] = (
    df["next_admittime"] - df["DISCHTIME"]
).dt.days

# Calculate difference in days between discharge and next admission for MIMIC-IV
df1["days_to_readmit1"] = (
    df1["next_admittime1"] - df1["dischtime"]
).dt.days

# Flag 30-day readmission for MIMIC-III
df["readmitted_30days"] = df["days_to_readmit"] <= 30

# Flag 30-day readmission for MIMIC-IV
df1["readmitted1_30days"] = df1["days_to_readmit1"] <= 30

# Patient-level readmission indicator for MIMIC-III
readmission_summary = (
    df.groupby("SUBJECT_ID")["readmitted_30days"]
      .any()
      .reset_index()
)

# Patient-level readmission indicator for MIMIC-IV
readmission_summary1 = (
    df1.groupby("subject_id")["readmitted1_30days"]
      .any()
      .reset_index()
)

# 5️⃣ MERGE ADMISSION + READMISSION RESULTS
# -------------------------------------------------------
final_df = admission_counts.merge(
    readmission_summary,
    on="SUBJECT_ID",
    how="left"
)

final_df1 = admission_counts1.merge(
    readmission_summary1,
    on="subject_id",
    how="left"
)

# Replace missing values with False
final_df["readmitted_30days"] = final_df["readmitted_30days"].fillna(False)
final_df1["readmitted1_30days"] = final_df1["readmitted1_30days"].fillna(False)

# -------------------------------------------------------
# 6️⃣ VISUALIZATION: ADMISSION DISTRIBUTION
# -------------------------------------------------------
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1) # Subplot for MIMIC-III
plt.hist(final_df["T_hosp_visits"], bins=20, color='skyblue', edgecolor='black')
plt.xlabel("Number of Hospital Visits")
plt.ylabel("Number of Patients")
plt.title("Distribution of Hospital Admissions (MIMIC-III)")

plt.subplot(1, 2, 2) # Subplot for MIMIC-IV
plt.hist(final_df1["T_hosp1_visits"], bins=20, color='lightcoral', edgecolor='black')
plt.xlabel("Number of Hospital Visits")
plt.ylabel("Number of Patients")
plt.title("Distribution of Hospital Admissions (MIMIC-IV)")

plt.tight_layout()
plt.show()

# -------------------------------------------------------
# 7️⃣ APA-STYLE SUMMARY TABLE
# -------------------------------------------------------
total_patients_mimic3 = final_df.shape[0]
total_patients_mimic4 = final_df1.shape[0]
mean_visits_mimic3 = final_df["T_hosp_visits"].mean()
mean_visits_mimic4 = final_df1["T_hosp1_visits"].mean()
sd_visits_mimic3 = final_df["T_hosp_visits"].std()
sd_visits_mimic4 = final_df1["T_hosp1_visits"].std()
readmission_rate_mimic3 = final_df["readmitted_30days"].mean() * 100
readmission_rate_mimic4 = final_df1["readmitted1_30days"].mean() * 100

apa_table = pd.DataFrame({
    "Measure": [
        "Total Patients (N)",
        "Mean Hospital Visits (SD)",
        "30-Day Readmission Rate (%)"
    ],
    "MIMIC-III Value": [
        total_patients_mimic3,
        f"{mean_visits_mimic3:.2f} ({sd_visits_mimic3:.2f})",
        f"{readmission_rate_mimic3:.2f}%"
    ],
    "MIMIC-IV Value": [
        total_patients_mimic4,
        f"{mean_visits_mimic4:.2f} ({sd_visits_mimic4:.2f})",
        f"{readmission_rate_mimic4:.2f}%"
    ]
})

print("\nAPA-Style Summary Table")
print(apa_table.to_string(index=False))

# -------------------------------------------------------
# 8️⃣ PREVIEW FINAL DATASET (FOR MODELING)
# -------------------------------------------------------
print("\nPreview of MIMIC-IV Modeling Dataset:")
print(final_df.head(20))
print("\nPreview of MIMIC-IV Modeling Dataset:")
print(final_df1.head(20))

# Saving the final results to a CSV file
final_df.to_csv("patient_admission_summary_mimic3.csv", index=False)
final_df1.to_csv("patient_admission_summary_mimic4.csv", index=False)
```

```python
# MIMIC-III Data Analysis
# =========================================================

import pandas as pd
import matplotlib.pyplot as plt

# MIMIC-III ONLY CONFIG (CSV files)
# =========================================================
adms_path = "/content/admissions.csv"                        # MIMIC-III admissions.csv
icu_path   = "/content/icustays.csv"                          # MIMIC-III icustays.csv

# Output CSV (final tabular result)
OUTPUT_CSV_PATH = "mimiciii_patient_adm_icu_readmission_summary.csv"

# 1) LOAD DATA (only required columns)
# =========================================================
# Admissions: subject_id, hadm_id, admittime, dischtime
adm = pd.read_csv(
    adms_path,
    usecols=["SUBJECT_ID", "HADM_ID", "ADMITTIME", "DISCHTIME"]
)
adm = adm.rename(columns={
    "SUBJECT_ID": "subject_id",
    "HADM_ID": "hadm_id",
    "ADMITTIME": "admittime",
    "DISCHTIME": "dischtime"
})
# ICU stays (MIMIC-III): subject_id, icustay_id (hadm_id optional but not required here)
icu = pd.read_csv(
    icu_path,
    usecols=["SUBJECT_ID", "ICUSTAY_ID"]
)

icu = icu.rename(columns={
    "SUBJECT_ID": "subject_id",
    "HADM_ID": "hadm_id",
    "ICUSTAY_ID": "icustay_id"
})

# 2) CLEAN / STANDARDIZE
# =========================================================
# Convert timestamps
adm["admittime"] = pd.to_datetime(adm["admittime"], errors="coerce")
adm["dischtime"] = pd.to_datetime(adm["dischtime"], errors="coerce")

# Drop rows missing keys or timestamps
adm = adm.dropna(subset=["subject_id", "hadm_id", "admittime", "dischtime"])
icu = icu.dropna(subset=["subject_id", "icustay_id"])

# Ensure subject_id behaves consistently as integer IDs
adm["subject_id"] = pd.to_numeric(adm["subject_id"], errors="coerce")
icu["subject_id"] = pd.to_numeric(icu["subject_id"], errors="coerce")
adm = adm.dropna(subset=["subject_id"])
icu = icu.dropna(subset=["subject_id"])
adm["subject_id"] = adm["subject_id"].astype("int64")
icu["subject_id"] = icu["subject_id"].astype("int64")

# =========================================================
# 3) ADMISSIONS: VISITS PER PATIENT + AVERAGE + FREQUENT FLYERS
# =========================================================
# Hospital visits = unique hadm_id per patient
admission_counts = (
    adm.groupby("subject_id")["hadm_id"]
       .nunique()
       .reset_index(name="total_hospital_visits")
)

# Frequent flyers: >= 3 hospital admissions
admission_counts["frequent_flyer_admissions_ge3"] = admission_counts["total_hospital_visits"] >= 3

# Average number of hospital visits per patient
avg_hospital_visits = admission_counts["total_hospital_visits"].mean()

# =========================================================
# 4) 30-DAY READMISSION (MIMIC-III)
#    Readmission event: next admission within 30 days after discharge
# =========================================================
adm_sorted = adm.sort_values(["subject_id", "admittime"]).copy()

# Next admission time per patient
adm_sorted["next_admittime"] = adm_sorted.groupby("subject_id")["admittime"].shift(-1)

# Days from discharge to next admission
adm_sorted["days_to_readmit"] = (adm_sorted["next_admittime"] - adm_sorted["dischtime"]).dt.days

# Event-level flag (ensure next admission is after discharge and within 30 days)
adm_sorted["readmit_30days_event"] = (
    adm_sorted["days_to_readmit"].notna()
    & (adm_sorted["days_to_readmit"] >= 0)
    & (adm_sorted["days_to_readmit"] <= 30)
)

# Patient-level: did they ever have a 30-day readmission?
readmission_patient = (
    adm_sorted.groupby("subject_id")["readmit_30days_event"]
              .any()
              .reset_index(name="readmitted_30days")
)

readmission_rate_pct = 100 * readmission_patient["readmitted_30days"].mean()

# =========================================================
# 5) ICU STAYS (MIMIC-III): ICU VISITS PER PATIENT + AVERAGE + FREQUENT FLYERS
# =========================================================
icu_counts = (
    icu.groupby("subject_id")["icustay_id"]
       .nunique()
       .reset_index(name="total_icu_visits")
)

# Frequent flyers: >= 3 ICU stays (optional but useful)
icu_counts["frequent_flyer_icu_ge3"] = icu_counts["total_icu_visits"] >= 3

avg_icu_visits = icu_counts["total_icu_visits"].mean()

# =========================================================
# 6) COMMON PATIENTS IN BOTH ADMISSIONS AND ICU + MERGE READMISSION
# =========================================================
# Keep patients who appear in BOTH hospital admissions and ICU stays
final_output = admission_counts.merge(icu_counts, on="subject_id", how="inner")
final_output["present_in_both_admissions_and_icustays"] = True

# Add readmission flag (patient-level)
final_output = final_output.merge(readmission_patient, on="subject_id", how="left")
final_output["readmitted_30days"] = final_output["readmitted_30days"].fillna(False)

# =========================================================
# 7) VISUALIZATIONS (MIMIC-III)
# =========================================================
plt.figure()
plt.hist(final_output["total_hospital_visits"], bins=20)
plt.xlabel("Number of Hospital Visits (Admissions)")
plt.ylabel("Number of Patients")
plt.title("MIMIC-III: Distribution of Hospital Admissions Per Patient")
plt.show()

plt.figure()
plt.hist(final_output["total_icu_visits"], bins=20)
plt.xlabel("Number of ICU Visits (ICU stays)")
plt.ylabel("Number of Patients")
plt.title("MIMIC-III: Distribution of ICU Stays Per Patient")
plt.show()

# =========================================================
# 8) APA-STYLE SUMMARY TABLE (MIMIC-III)
# =========================================================
total_admitted_patients = admission_counts["subject_id"].nunique()
total_icu_patients = icu_counts["subject_id"].nunique()
total_common_patients = final_output["subject_id"].nunique()

mean_visits = admission_counts["total_hospital_visits"].mean()
sd_visits = admission_counts["total_hospital_visits"].std()

mean_icu = icu_counts["total_icu_visits"].mean()
sd_icu = icu_counts["total_icu_visits"].std()

freq_flyers_adm_n = int(admission_counts["frequent_flyer_admissions_ge3"].sum())
freq_flyers_adm_pct = 100 * admission_counts["frequent_flyer_admissions_ge3"].mean()

freq_flyers_icu_n = int(icu_counts["frequent_flyer_icu_ge3"].sum())
freq_flyers_icu_pct = 100 * icu_counts["frequent_flyer_icu_ge3"].mean()

apa_table = pd.DataFrame({
    "Measure": [
        "Patients with ≥1 hospital admission (N)",
        "Patients with ≥1 ICU stay (N)",
        "Patients in BOTH admissions & ICU (N)",
        "Mean hospital visits per patient (SD)",
        "Mean ICU visits per patient (SD)",
        "Frequent flyers: ≥3 admissions (N, %)",
        "Frequent flyers: ≥3 ICU stays (N, %)",
        "30-day readmission rate (%, patient-level)"
    ],
    "Value": [
        f"{total_admitted_patients}",
        f"{total_icu_patients}",
        f"{total_common_patients}",
        f"{mean_visits:.2f} ({sd_visits:.2f})",
        f"{mean_icu:.2f} ({sd_icu:.2f})",
        f"{freq_flyers_adm_n} ({freq_flyers_adm_pct:.2f}%)",
        f"{freq_flyers_icu_n} ({freq_flyers_icu_pct:.2f}%)",
        f"{readmission_rate_pct:.2f}%"
    ]
})

print("\n================ MIMIC-III APA-STYLE SUMMARY TABLE ================")
print(apa_table.to_string(index=False))

# =========================================================
# 9) FINAL TABULAR OUTPUT (MODEL-READY) + EXPORT TO CSV
# =========================================================
# Clean, modeling-friendly columns
final_output = final_output[[
    "subject_id",
    "total_hospital_visits",
    "frequent_flyer_admissions_ge3",
    "total_icu_visits",
    "frequent_flyer_icu_ge3",
    "readmitted_30days",
    "present_in_both_admissions_and_icustays"
]].sort_values(
    by=["readmitted_30days", "frequent_flyer_admissions_ge3", "total_hospital_visits", "total_icu_visits"],
    ascending=[False, False, False, False]
).reset_index(drop=True)

print("\n================ MIMIC-III FINAL OUTPUT (TOP 15 ROWS) ================")
print(final_output.head(15).to_string(index=False))

# Export to CSV
final_output.to_csv(OUTPUT_CSV_PATH, index=False, encoding="utf-8")
print(f"\n✅ Exported final table to: {OUTPUT_CSV_PATH}")
```

```python

# MIMIC-IV Data Analysis
# =========================================================

import pandas as pd
import matplotlib.pyplot as plt

# MIMIC-IV ONLY CONFIG (CSV files)
# =========================================================
adms1_path = "/content/admissions1.csv"                        # MIMIC-IV admissions.csv
icu1_path   = "/content/icustays1.csv"                          # MIMIC-IV icustays.csv

# Output CSV (final tabular result)
OUTPUT_CSV_PATH1 = "mimicIV_patient_adm_icu_readmission_summary.csv"

# 1) LOAD DATA (only required columns)
# =========================================================
# Admissions: subject_id, hadm_id, admittime, dischtime
adm1 = pd.read_csv(
    adms1_path,
    usecols=["subject_id", "hadm_id", "admittime", "dischtime"]
)

# ICU stays (MIMIC-IV): subject_id, icustay_id (hadm_id optional but not required here)
icu1 = pd.read_csv(
    icu1_path,
    usecols=["subject_id", "stay_id"]
)

# 2) CLEAN / STANDARDIZE
# =========================================================
# Convert timestamps
adm1["admittime"] = pd.to_datetime(adm1["admittime"], errors="coerce")
adm1["dischtime"] = pd.to_datetime(adm1["dischtime"], errors="coerce")

# Drop rows missing keys or timestamps
adm1 = adm1.dropna(subset=["subject_id", "hadm_id", "admittime", "dischtime"])
icu1 = icu1.dropna(subset=["subject_id", "stay_id"])

# Ensure subject_id behaves consistently as integer IDs
adm1["subject_id"] = pd.to_numeric(adm1["subject_id"], errors="coerce")
icu1["subject_id"] = pd.to_numeric(icu1["subject_id"], errors="coerce")
adm1 = adm1.dropna(subset=["subject_id"])
icu1 = icu1.dropna(subset=["subject_id"])
adm1["subject_id"] = adm1["subject_id"].astype("int64")
icu1["subject_id"] = icu1["subject_id"].astype("int64")

# =========================================================
# 3) ADMISSIONS: VISITS PER PATIENT + AVERAGE + FREQUENT FLYERS
# =========================================================
# Hospital visits = unique hadm_id per patient
admission_counts = (
    adm1.groupby("subject_id")["hadm_id"]
       .nunique()
       .reset_index(name="total_hospital_visits")
)

# Frequent flyers: >= 3 hospital admissions
admission_counts["frequent_flyer_admissions_ge3"] = admission_counts["total_hospital_visits"] >= 3

# Average number of hospital visits per patient
avg_hospital_visits = admission_counts["total_hospital_visits"].mean()

# =========================================================
# 4) 30-DAY READMISSION (MIMIC-IV)
#    Readmission event: next admission within 30 days after discharge
# =========================================================
adm_sorted = adm1.sort_values(["subject_id", "admittime"]).copy()

# Next admission time per patient
adm_sorted["next_admittime"] = adm_sorted.groupby("subject_id")["admittime"].shift(-1)

# Days from discharge to next admission
adm_sorted["days_to_readmit"] = (adm_sorted["next_admittime"] - adm_sorted["dischtime"]).dt.days

# Event-level flag (ensure next admission is after discharge and within 30 days)
adm_sorted["readmit_30days_event"] = (
    adm_sorted["days_to_readmit"].notna()
    & (adm_sorted["days_to_readmit"] >= 0)
    & (adm_sorted["days_to_readmit"] <= 30)
)

# Patient-level: did they ever have a 30-day readmission?
readmission_patient = (
    adm_sorted.groupby("subject_id")["readmit_30days_event"]
              .any()
              .reset_index(name="readmitted_30days")
)

readmission_rate_pct = 100 * readmission_patient["readmitted_30days"].mean()

# =========================================================
# 5) ICU STAYS (MIMIC-IV): ICU VISITS PER PATIENT + AVERAGE + FREQUENT FLYERS
# =========================================================
icu_counts = (
    icu1.groupby("subject_id")["stay_id"]
       .nunique()
       .reset_index(name="total_icu_visits")
)

# Frequent flyers: >= 3 ICU stays (optional but useful)
icu_counts["frequent_flyer_icu_ge3"] = icu_counts["total_icu_visits"] >= 3

avg_icu_visits = icu_counts["total_icu_visits"].mean()

# =========================================================
# 6) COMMON PATIENTS IN BOTH ADMISSIONS AND ICU + MERGE READMISSION
# =========================================================
# Keep patients who appear in BOTH hospital admissions and ICU stays
final_output = admission_counts.merge(icu_counts, on="subject_id", how="inner")
final_output["present_in_both_admissions_and_icustays"] = True

# Add readmission flag (patient-level)
final_output = final_output.merge(readmission_patient, on="subject_id", how="left")
final_output["readmitted_30days"] = final_output["readmitted_30days"].fillna(False)

# =========================================================
# 7) VISUALIZATIONS (MIMIC-IV)
# =========================================================
plt.figure()
plt.hist(final_output["total_hospital_visits"], bins=20)
plt.xlabel("Number of Hospital Visits (Admissions)")
plt.ylabel("Number of Patients")
plt.title("MIMIC-IV: Distribution of Hospital Admissions Per Patient")
plt.show()

plt.figure()
plt.hist(final_output["total_icu_visits"], bins=20)
plt.xlabel("Number of ICU Visits (ICU stays)")
plt.ylabel("Number of Patients")
plt.title("MIMIC-IV: Distribution of ICU Stays Per Patient")
plt.show()

# =========================================================
# 8) APA-STYLE SUMMARY TABLE (MIMIC-IV)
# =========================================================
total_admitted_patients = admission_counts["subject_id"].nunique()
total_icu_patients = icu_counts["subject_id"].nunique()
total_common_patients = final_output["subject_id"].nunique()

mean_visits = admission_counts["total_hospital_visits"].mean()
sd_visits = admission_counts["total_hospital_visits"].std()

mean_icu = icu_counts["total_icu_visits"].mean()
sd_icu = icu_counts["total_icu_visits"].std()

freq_flyers_adm_n = int(admission_counts["frequent_flyer_admissions_ge3"].sum())
freq_flyers_adm_pct = 100 * admission_counts["frequent_flyer_admissions_ge3"].mean()

freq_flyers_icu_n = int(icu_counts["frequent_flyer_icu_ge3"].sum())
freq_flyers_icu_pct = 100 * icu_counts["frequent_flyer_icu_ge3"].mean()

apa_table = pd.DataFrame({
    "Measure": [
        "Patients with ≥1 hospital admission (N)",
        "Patients with ≥1 ICU stay (N)",
        "Patients in BOTH admissions & ICU (N)",
        "Mean hospital visits per patient (SD)",
        "Mean ICU visits per patient (SD)",
        "Frequent flyers: ≥3 admissions (N, %)",
        "Frequent flyers: ≥3 ICU stays (N, %)",
        "30-day readmission rate (%, patient-level)"
    ],
    "Value": [
        f"{total_admitted_patients}",
        f"{total_icu_patients}",
        f"{total_common_patients}",
        f"{mean_visits:.2f} ({sd_visits:.2f})",
        f"{mean_icu:.2f} ({sd_icu:.2f})",
        f"{freq_flyers_adm_n} ({freq_flyers_adm_pct:.2f}%)",
        f"{freq_flyers_icu_n} ({freq_flyers_icu_pct:.2f}%)",
        f"{readmission_rate_pct:.2f}%"
    ]
})

print("\n================ MIMIC-IV APA-STYLE SUMMARY TABLE ================")
print(apa_table.to_string(index=False))

# =========================================================
# 9) FINAL TABULAR OUTPUT (MODEL-READY) + EXPORT TO CSV
# =========================================================
# Clean, modeling-friendly columns
final_output = final_output[[
    "subject_id",
    "total_hospital_visits",
    "frequent_flyer_admissions_ge3",
    "total_icu_visits",
    "frequent_flyer_icu_ge3",
    "readmitted_30days",
    "present_in_both_admissions_and_icustays"
]].sort_values(
    by=["readmitted_30days", "frequent_flyer_admissions_ge3", "total_hospital_visits", "total_icu_visits"],
    ascending=[False, False, False, False]
).reset_index(drop=True)

print("\n================ MIMIC-IV FINAL OUTPUT (TOP 15 ROWS) ================")
print(final_output.head(15).to_string(index=False))

# Export to CSV
final_output.to_csv(OUTPUT_CSV_PATH1, index=False, encoding="utf-8")
print(f"\n✅ Exported final table to: {OUTPUT_CSV_PATH1}")
```
